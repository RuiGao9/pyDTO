```python
import pyDTO

def test_run():
    # 测试坐标：旧金山 (San Francisco)
    # 这里的 West 距离应该是你最关心的“到海洋距离”
    sf_lon, sf_lat = -122.4194, 37.7749
    
    print("--- Running pyDTO Test ---")
    result = pyDTO.get_dist(sf_lon, sf_lat, unit='km')
    
    if "error" in result:
        print(f"Failed: {result['error']}")
    else:
        dist = result['distances']
        print(f"State: California")
        print(f"Distance to West (Ocean): {dist['West']} km")
        print(f"Distance to East: {dist['East']} km")
        print(f"Distance to North: {dist['North']} km")
        print(f"Distance to South: {dist['South']} km")

if __name__ == "__main__":
    test_run()
```
